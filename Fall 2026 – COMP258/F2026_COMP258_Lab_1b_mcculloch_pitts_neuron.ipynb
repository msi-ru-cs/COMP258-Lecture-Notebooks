# Mcculloch_Pitts Neuron

```python
def mcculloch_pitts_neuron(inputs, weights, threshold):
    """Simple McCulloch-Pitts neuron model."""
    weighted_sum = sum(i * w for i, w in zip(inputs, weights))
    return 1 if weighted_sum >= threshold else 0


# Pattern for upscale detection: do -> re -> mi
def detect_upscale(pattern_t1, pattern_t2, pattern_t3):
    weights = [1, 1, 1]
    threshold = 3
    # Create a combined pattern based on time steps
    combined_pattern = [pattern_t1[0], pattern_t2[1], pattern_t3[2]]
    return mcculloch_pitts_neuron(combined_pattern, weights, threshold)

# Pattern for downscale detection: mi -> re -> do
def detect_downscale(pattern_t1, pattern_t2, pattern_t3):
    weights = [1, 1, 1]
    threshold = 3
    # Create a combined pattern based on reverse order for downscale
    combined_pattern = [pattern_t1[2], pattern_t2[1], pattern_t3[0]]
    return mcculloch_pitts_neuron(combined_pattern, weights, threshold)

```

```python
# Define time steps for the upscale and downscale patterns
# do = 1, re = 2, mi = 3
do = [1, 0, 0]  # Input pattern at t=1
re = [0, 1, 0]  # Input pattern at t=2
mi = [0, 0, 1]  # Input pattern at t=3
```

```python
# Test upscale segment detection
print("Upscale Detection (do -> re -> mi):")
upscale_output = detect_upscale(do, re, mi)
print(f"Upscale Detected: {upscale_output}")

# Test downscale segment detection
print("Downscale Detection (mi -> re -> do):")
downscale_output = detect_downscale(mi, re, do)
print(f"Downscale Detected: {downscale_output}")

# Test upscale segment detection
print("Upscale Detection (mi -> re -> do):")
upscale_output = detect_upscale(re, mi, do)
print(f"Upscale Detected: {upscale_output}")
```
