# Perceptron

```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
# AND
input_data = [(0, 0), (0, 1), (1, 0), (1, 1)]
output_data = [0, 0, 0, 1]
input_data, output_data
```

```python
# Initialize weight and bias
w1 = random.uniform(-1, 1)
w2 = random.uniform(-1, 1)
b = random.uniform(-1, 1)
w1, w2, b
```

```python
# Learning rate
lr = 0.01
# Trials
epochs = 100
```

```python
# Step activation function
def activate(weighted_sum):
  threshold = 0
  if weighted_sum >= threshold:
    return 1
  else:
    return 0
```

```python
for epoch in range(epochs):
  print(f"\nEpoch: {epoch}")
  total_error = 0

  for i in range(len(input_data)):
    x1, x2 = input_data[i]
    y = output_data[i]
    print(f"x1: {x1}, x2: {x2}, y: {y}")

    # Compute weighted sum
    weighted_sum = x1 * w1 + x2 * w2 + b
    print(f"weighted_sum: {weighted_sum}")

    # Compute activation
    y_pred = activate(weighted_sum)
    print(f"prediction (y_hat): {y_pred}")

    # Compute error
    error = y - y_pred
    print(f"error: {error}")

    # Update weights and bias
    w1 = w1 + lr * error * x1
    w2 = w2 + lr * error * x2
    b = b + lr * error
    print(f"updated weights: w1: {w1}, w2: {w2}, b: {b}")

    total_error += abs(error)
    print(f"total error: {total_error}")

  if total_error == 0:
    print("Model converged at epoch: ", epoch)
    break






```
