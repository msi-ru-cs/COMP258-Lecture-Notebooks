```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Derivatives of activation functions
def sigmoid_derivative(x):
    return sigmoid(x) * (1 - sigmoid(x))

def tanh_derivative(x):
    return 1 - np.tanh(x)**2

def relu_derivative(x):
    return np.where(x > 0, 1, 0)
```

```python
# Generate x values
x_values = np.linspace(-5, 5, 100)

# Calculate y values for each activation function and their derivatives
sigmoid_values = sigmoid(x_values)
tanh_values = tanh(x_values)
relu_values = relu(x_values)

sigmoid_derivative_values = sigmoid_derivative(x_values)
tanh_derivative_values = tanh_derivative(x_values)
relu_derivative_values = relu_derivative(x_values)
```

```python
# Plot the activation functions
plt.figure(figsize=(10, 6))

# # Sigmoid
plt.plot(x_values, sigmoid_values, label='Sigmoid', color='blue')
plt.plot(x_values, sigmoid_derivative_values, label='Sigmoid Derivative', linestyle='--', color='blue')

 # # Tanh
plt.plot(x_values, tanh_values, label='Tanh', color='orange')
plt.plot(x_values, tanh_derivative_values, label='Tanh Derivative', linestyle='--', color='orange')

# # # # # # ReLU
plt.plot(x_values, relu_values, label='ReLU', color='green')
plt.plot(x_values, relu_derivative_values, label='ReLU Derivative', linestyle='--', color='green')

# Set labels and title
plt.title('Activation Functions and Their Derivatives')
plt.xlabel('Input')
plt.ylabel('Output')
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.grid(color = 'gray', linestyle = '--', linewidth = 0.5)
plt.legend()

# Show the plot
plt.show()
```
